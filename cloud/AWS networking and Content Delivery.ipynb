## EL MODELO OSI
1) Físico: transmisión y recepción por un medio físico de bitstreams (1 or 0)
2) Enlace de datos: transferencia de datos en una red LAN (hubs and switches) MAC
3) Red: rutea y empaqueta (routers) IP
4) Transporte: brinda protocolos de soporte para la comunicación host to host (TCP, UDP)
5) Sesión: permite el intercambio ordenado de datos (NetBIOS, RPC)
6) Presentación: asegura que la aplicación pueda leer los datos; encriptación (ASCI, ICA)
7) Aplicación: acceso de aplicaciones a red de computadoras (HTTPS, FTP, DHCP, LDAP)
